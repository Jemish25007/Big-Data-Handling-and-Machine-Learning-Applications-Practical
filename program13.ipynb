```python

# Raw text data
text = "  HELLO!!!   Welcome, TO Python @ Data Science.  "

print("Original Text:", text)

# Convert text to lowercase
text = text.lower()

# Remove special punctuation symbols
text = text.replace("!", "")
text = text.replace(",", "")
text = text.replace("@", "")
text = text.replace(".", "")

# Remove extra spaces from beginning and end
text = text.strip()

# Split text into individual words
words = text.split()

print("Cleaned Text:", text)
print("Word Tokens:", words)
```
