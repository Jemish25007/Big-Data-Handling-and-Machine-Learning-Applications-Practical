```python

from pyspark.sql import SparkSession

# 1. Create Spark session
spark = SparkSession.builder \
    .appName("ParquetExample") \
    .master("local[*]") \
    .getOrCreate()

# 2. Create DataFrame
data = [
    (1, "Amit", 20000),
    (2, "Rahul", 25000),
    (3, "Priya", 30000)
]

columns = ["ID", "Name", "Salary"]

df = spark.createDataFrame(data, columns)

print("Original Data:")
df.show()

# 3. Save DataFrame as Parquet
df.write.mode("overwrite").parquet("employee_parquet")

# 4. Read Parquet file
df2 = spark.read.parquet("employee_parquet")

print("Loaded Parquet Data:")
df2.show()

# 5. Verify schema
print("Original Schema:")
df.printSchema()

print("Parquet Schema:")
df2.printSchema()

# 6. Verify row counts
print("Original Row Count:", df.count())
print("Parquet Row Count:", df2.count())

# 7. Stop Spark
spark.stop()
```
