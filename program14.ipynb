```python

import re
from collections import Counter

# Text corpus
text = """
Python is easy to learn and Python is useful.
Data science is interesting and data science is growing.
Python and data are important in data science.
"""

# Convert to lowercase
text = text.lower()

# Remove punctuation and special symbols
text = re.sub(r'[^a-z\s]', '', text)

# Split text into words
words = text.split()

# Common English stopwords
stopwords = ['the', 'is', 'in', 'and', 'a', 'to', 'of']

# Remove stopwords
keywords = [word for word in words if word not in stopwords]

# Count word frequencies
frequency = Counter(keywords)

# Display top 5 keywords
print("Top 5 Keywords:")
print(frequency.most_common(5))
```
