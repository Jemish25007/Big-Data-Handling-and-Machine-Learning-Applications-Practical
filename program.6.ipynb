```python

from pyspark.sql import SparkSession
from pyspark.sql.functions import col

# 1. Create Spark session
spark = SparkSession.builder.appName("DataTransform").master("local[*]").getOrCreate()

# 2. Create sample data
data = [
    (1, "Amit", 65000, "Surat"),
    (2, "Rahul", 50000, "Ahmedabad"),
    (3, "Priya", 70000, "Surat"),
    (4, "Neha", 80000, "Vadodara")
]

columns = ["ID", "Name", "Salary", "City"]

df = spark.createDataFrame(data, columns)

print("Original Data:")
df.show()

# 3. Select specific columns
df.select("Name", "Salary", "City").show()

# 4. Filter rows using two conditions
df.filter((col("Salary") > 60000) & (col("City") == "Surat")).show()

# 5. Create a computed column (10% bonus)
df = df.withColumn("Bonus", col("Salary") * 0.10)

# 6. Rename a column
df = df.withColumnRenamed("Name", "Employee_Name")

print("Final Data:")
df.show()

spark.stop()
```
