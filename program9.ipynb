```python

import pandas as pd
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import LabelEncoder, OneHotEncoder

# Create sample dataset
data = {
    'Age': [20, None, 25, 30],
    'Rating': [4.5, 3.5, None, 5.0],
    'Level': ['Low', 'Medium', 'High', 'Low'],
    'City': ['Surat', 'Mumbai', 'Surat', 'Delhi']
}

df = pd.DataFrame(data)

print("Original Data:")
print(df)

# 1. Fill missing numerical values with mean
imputer = SimpleImputer(strategy='mean')
df[['Age', 'Rating']] = imputer.fit_transform(df[['Age', 'Rating']])

# 2. Convert text labels into integers
le = LabelEncoder()
df['Level'] = le.fit_transform(df['Level'])

# 3. Convert nominal categories into binary columns
encoder = OneHotEncoder(sparse_output=False, handle_unknown='ignore')
city_encoded = encoder.fit_transform(df[['City']])

city_df = pd.DataFrame(
    city_encoded,
    columns=encoder.get_feature_names_out(['City'])
)

# Combine encoded columns with dataset
df = pd.concat([df.drop(columns=['City']), city_df], axis=1)

print("\nPreprocessed Data:")
print(df)
```
