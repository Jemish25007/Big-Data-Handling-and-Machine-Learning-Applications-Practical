PRACTICAL 05: Apache Spark Session Initialization & DataFrame Schema Inspection

```python

from pyspark.sql import SparkSession

# 1. Create Spark session
spark = SparkSession.builder \
    .appName("MyFirstSpark") \
    .master("local[*]") \
    .getOrCreate()

# 2. Create data using dictionaries
data = [
    {"Name": "Amit", "Age": 20, "City": "Surat"},
    {"Name": "Rahul", "Age": 21, "City": "Ahmedabad"},
    {"Name": "Priya", "Age": 22, "City": "Vadodara"}
]

# 3. Create PySpark DataFrame
df = spark.createDataFrame(data)

# 4. Display schema
df.printSchema()

# 5. Display rows
df.show()

# 6. Stop Spark session
spark.stop()
```
