PRACTICAL 04: Out-of-Core Batch Ingestion & Aggregation via Chunking

```python

import pandas as pd


total_records = 0
total_year = 0
valid_years = 0

for chunk in pd.read_csv("netflix_titles.csv", chunksize=1000):

    
    total_records += len(chunk)

    
    years = pd.to_numeric(chunk['release_year'], errors='coerce')
    total_year += years.sum()
    valid_years += years.count()


if valid_years > 0:
    average = total_year / valid_years
else:
    average = 0

print("Total Records:", total_records)
print("Sum of Release Years:", total_year)
print("Average Release Year:", round(average, 2))
```
