```python

from pyspark.sql import SparkSession
from pyspark.sql.functions import count, avg, max, sum

# 1. Create Spark session
spark = SparkSession.builder.appName("GroupAggregation").master("local[*]").getOrCreate()

# 2. Create sample data
data = [
    ("IT", 50000),
    ("IT", 60000),
    ("HR", 40000),
    ("HR", 45000),
    ("Sales", 55000)
]

columns = ["Department", "Salary"]

df = spark.createDataFrame(data, columns)

print("Original Data:")
df.show()

# 3. Group data and calculate aggregates
result = df.groupBy("Department").agg(
    count("*").alias("Employee_Count"),
    avg("Salary").alias("Average_Salary"),
    max("Salary").alias("Maximum_Salary"),
    sum("Salary").alias("Total_Salary")
)

# 4. Sort by total salary in descending order
result = result.orderBy("Total_Salary", ascending=False)

print("Department Summary:")
result.show()

# 5. Count the number of departments
print("Total Departments:", result.count())

# 6. Stop Spark
spark.stop()
```
